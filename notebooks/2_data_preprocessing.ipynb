In this notebook, I will produce the training/test data for ARIMAX, FFNN, RNN

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
```

```python
from utils.preprocessing import generate_data_ffnn
import numpy as np
import pandas as pd
```

```python
cpi = pd.read_csv('..\\data\\structured_data\\CPI_growth_yoy_dataframe.csv')
rgdp = pd.read_csv('..\\data\\structured_data\\RGDP_growth_dataframe.csv')
unemp = pd.read_csv('..\\data\\structured_data\\unemployment_rate_dataframe.csv')
# dropping the unwanted duplicate columns and indexes, renaming the observed value columns
cpi = cpi.drop(['Unnamed: 0'], axis = 1).rename(columns = {'OBS_VALUE': 'CPI inflation'})
rgdp = rgdp.drop(['Unnamed: 0', 'Reference area', 'TIME_PERIOD'], axis = 1).rename(columns = {'OBS_VALUE': 'RGDP growth'})
unemp = unemp.drop(['Unnamed: 0', 'Reference area', 'TIME_PERIOD'], axis = 1).rename(columns = {'OBS_VALUE': 'Unemp rate'})
```

Generating the training and testing inputs/targets for FFNN

```python
ffnn_df = pd.concat((cpi, rgdp, unemp), axis = 1)
# dropping the country name and timestep indexes
ffnn_arr = ffnn_df.to_numpy()[:, 2:]
ffnn_arr.shape
```

```python
lags = 3
ffnn_train_inputs, ffnn_train_targets = generate_data_ffnn(array = ffnn_arr, number_of_lags = lags, num_countries = 22, num_timesteps = 104, num_features = 3)
ffnn_train_inputs.shape
```

```python
np.save('..\\data\\input_data\\ffnn_train_inputs.npy', ffnn_train_inputs)
np.save('..\\data\\input_data\\ffnn_train_targets.npy', ffnn_train_targets)
a = np.load('..\\data\\input_data\\ffnn_train_inputs.npy', allow_pickle = True)
a.shape
```
