The years 1994-2009 will be used for training, years 2010-2014 for evaluation during the inital training process to determine the most suitable hyperparameters' configuration, and years 2015-2019 for rolling-origin forecasting to produce the forecast errors

```python
cpi = pd.read_csv('..\\data\\structured_data\\CPI_growth_yoy_dataframe.csv')
rgdp = pd.read_csv('..\\data\\structured_data\\RGDP_growth_dataframe.csv')
unemp = pd.read_csv('..\\data\\structured_data\\unemployment_rate_dataframe.csv')
# dropping the unwanted duplicate columns and indexes, renaming the observed value columns
cpi = cpi.drop(['Unnamed: 0'], axis = 1).rename(columns = {'OBS_VALUE': 'CPI inflation'})
rgdp = rgdp.drop(['Unnamed: 0', 'Reference area', 'TIME_PERIOD'], axis = 1).rename(columns = {'OBS_VALUE': 'RGDP growth'})
unemp = unemp.drop(['Unnamed: 0', 'Reference area', 'TIME_PERIOD'], axis = 1).rename(columns = {'OBS_VALUE': 'Unemp rate'})

train_timesteps_list = generate_timesteps_list(1994, 2009)
val_timesteps_list = generate_timesteps_list(2010, 2014)
test_timesteps_list = generate_timesteps_list(2015, 2019)

# concatenating along the columns
df = pd.concat((cpi, rgdp, unemp), axis = 1)

train_df = df.loc[df['TIME_PERIOD'].isin(train_timesteps_list), :]
# dropping the country name and timestep indexes
train_arr = train_df.to_numpy()[:, 2:]

val_df = df.loc[df['TIME_PERIOD'].isin(val_timesteps_list), :]
val_arr = val_df.to_numpy()[:, 2:]

test_df = df.loc[df['TIME_PERIOD'].isin(test_timesteps_list), :]
test_arr = test_df.to_numpy()[:, 2:]
```
