This notebook explores the performance of each of the model, comparing their MAE and RMSE, as well as running a Diebold-Mariano test on MAE, which is first averaged across countries for each timestep for each model. Furthermore, the performance of models on the indicators' values during and post COVID recession is also evaluated 

```python
import pandas as pd
import numpy as np
from pathlib import Path
import sys
sys.path.append(str(Path.cwd().parent))
import matplotlib.pyplot as plt
from utils.misc import produce_arimax_orders_csv
from utils.tests import diebold_mariano, MAE, RMSE, produce_errors_per_country_csv, produce_errors_across_countries_csv
```

```python
naive_df = pd.read_csv('..\\data\\forecasts\\naive_forecast_dataframe.csv').drop(['Unnamed: 0'], axis = 1)
ffnn_df = pd.read_csv('..\\data\\forecasts\\ffnn_forecasts_dataframe.csv').drop(['Unnamed: 0'], axis = 1)
rnn_df = pd.read_csv('..\\data\\forecasts\\rnn_forecasts_dataframe.csv').drop(['Unnamed: 0'], axis = 1)
arimax_df = pd.read_csv('..\\data\\forecasts\\arimax_forecasts_dataframe.csv').drop(['Unnamed: 0'], axis = 1)
# generating the final table with the forecasts for all the models
final_forecasts_df = naive_df.copy().rename(columns = {'Forecasts': 'Naive forecast'})
final_forecasts_df['FFNN'] = ffnn_df.iloc[:, -1]
final_forecasts_df['RNN'] = rnn_df.iloc[:, -1]
final_forecasts_df['ARIMAX'] = arimax_df.iloc[:, -1]
final_forecasts_df.to_csv('..\\results\\main_results\\forecasts_dataframe.csv', )
```

```python
filepath = '..\\results\\main_results\\DM_tests.csv'
diebold_mariano(naive_df, ffnn_df, rnn_df, arimax_df, filepath)
```

```python
print(f'MAE for naive forecasts is {MAE(naive_df)}')
print(f'MAE for FFNN forecasts is {MAE(ffnn_df)}')
print(f'MAE for RNN forecasts is {MAE(rnn_df)}')
print(f'MAE for ARIMAX forecasts is {MAE(arimax_df)}')
print(f'RMSE for naive forecasts is {RMSE(naive_df)}')
print(f'RMSE for FFNN forecasts is {RMSE(ffnn_df)}')
print(f'RMSE for RNN forecasts is {RMSE(rnn_df)}')
print(f'RMSE for ARIMAX forecasts is {RMSE(arimax_df)}')
filepath = '..\\results\\main_results\\errors_per_country.csv'
produce_errors_per_country_csv(naive_df, ffnn_df, rnn_df, arimax_df, filepath)
filepath = '..\\results\\main_results\\errors_across_countries.csv'
produce_errors_across_countries_csv(naive_df, ffnn_df, rnn_df, arimax_df, filepath)
```
